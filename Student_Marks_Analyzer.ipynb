```python

# Student Marks Analyzer Program

# Students data
students = [
    {"name": "Amit", "maths": 78, "science": 85, "english": 69},
    {"name": "Neha", "maths": 92, "science": 88, "english": 91},
    {"name": "Rahul", "maths": 60, "science": 55, "english": 58},
    {"name": "Pooja", "maths": 81, "science": 73, "english": 77}
]

# Variables initialization
max_marks = 0
topper = ""
class_total = 0
passed_students = 0

print("Name      Total   Average   Grade")

# Loop through each student
for student in students:
    total = student["maths"] + student["science"] + student["english"]
    average = total / 3

    # Grade calculation
    if average >= 85:
        grade = "A"
    elif average >= 70:
        grade = "B"
    elif average >= 50:
        grade = "C"
    else:
        grade = "F"

    # Topper check
    if total > max_marks:
        max_marks = total
        topper = student["name"]

    # Class total calculation
    class_total += total

    # Pass count
    if average >= 50:
        passed_students += 1

    # Print student result
    print(student["name"], "   ", total, "    ", int(average), "      ", grade)

# Final results
class_average = class_total / len(students)

print("\nTopper is:", topper, "with marks", max_marks)
print("Class average is:", class_average)
print("Total students passed:", passed_students)
```
