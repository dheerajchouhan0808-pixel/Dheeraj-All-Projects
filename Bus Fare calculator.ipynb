```python
names = ["Amit", "Neha", "Rahul", "Pooja"]
distances = [12, 25, 8, 18]

fares = []          # final fares store karne ke liye
total_fare = 0
total_distance = 0

# Task 1: Fare Calculation
for i in range(len(names)):
    distance = distances[i]
    base_fare = distance * 10

    if distance >= 20:
        discount = base_fare * 0.15
    elif distance >= 10:
        discount = base_fare * 0.10
    else:
        discount = 0

    final_fare = base_fare - discount
    fares.append(final_fare)

    total_fare += final_fare
    total_distance += distance

# Task 2: Display Output
print("Name   Distance(km)   Fare(₹)")
for i in range(len(names)):
    print(names[i], "   ", distances[i], "         ", int(fares[i]))

# Task 3: Daily Summary
highest_fare = max(fares)
index = fares.index(highest_fare)
highest_payer = names[index]

average_distance = total_distance / len(distances)

print("\n--- Daily Summary ---")
print("Total Fare Collected: ₹", int(total_fare))
print("Highest Fare Paid By:", highest_payer)
print("Average Distance Travelled:", average_distance, "km")

```
