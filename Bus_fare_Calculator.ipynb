```python
# Bus fare Calculator.                         # Program ka title

names = ["Amit", "Neha", "Rahul", "Pooja"]     # Passengers ke naam
distances = [12, 25, 8, 18]                   # Travel distances

fares = []                                    # Final fares list
total_fare = 0                                # Total fare sum
total_distance = 0                            # Total distance sum

# Task 1: Fare Calculation                     # Fare calculate karna
for i in range(len(names)):                   # Har passenger ke liye
    distance = distances[i]                   # Current distance
    base_fare = distance * 10                 # Base fare nikala

    if distance >= 20:                        # 20 km ya zyada
        discount = base_fare * 0.15           # 15% discount
    elif distance >= 10:                      # 10 km ya zyada
        discount = base_fare * 0.10           # 10% discount
    else:                                     # 10 km se kam
        discount = 0                          # No discount

    final_fare = base_fare - discount         # Final fare nikala
    fares.append(final_fare)                  # Fare list me add

    total_fare += final_fare                  # Total fare update
    total_distance += distance                # Total distance update

# Task 2: Display Output                        # Output dikhana
print("Name   Distance(km)   Fare(₹)")         # Heading print
for i in range(len(names)):                   # Dobara loop
    print(names[i], "   ", distances[i], "         ", int(fares[i]))

                                               # Details print

# Task 3: Daily Summary                         # Summary banana
highest_fare = max(fares)                     # Sabse bada fare
index = fares.index(highest_fare)             # Uska index
highest_payer = names[index]                  # Naam nikala

average_distance = total_distance / len(distances)
                                               # Average distance

print("\n--- Daily Summary ---")               # Summary title
print("Total Fare Collected: ₹", int(total_fare))
                                               # Total fare print
print("Highest Fare Paid By:", highest_payer) # Highest payer
print("Average Distance Travelled:", average_distance, "km")
                                               # Average distance
```
