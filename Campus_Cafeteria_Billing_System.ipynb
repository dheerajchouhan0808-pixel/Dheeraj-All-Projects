```python
# Campus Cafeteria Billing System.   # Program ka naam

menu = {                              # Menu aur price
    "Burger": 50,                     # Burger ka rate
    "Pizza": 100,                     # Pizza ka rate
    "Sandwich": 40,                   # Sandwich ka rate
    "Coffee": 30,                     # Coffee ka rate
    "Tea": 20                         # Tea ka rate
}

orders = [                            # Sabke orders list
    {"name": "Amit", "items": {"Burger": 2, "Coffee": 1}},   # Amit ka order
    {"name": "Neha", "items": {"Pizza": 1, "Sandwich": 2, "Tea": 2}}, # Neha ka order
    {"name": "Rahul", "items": {"Sandwich": 3, "Coffee": 2}}, # Rahul ka order
    {"name": "Pooja", "items": {"Burger": 1, "Pizza": 1, "Coffee": 1}} # Pooja ka order
]

highest_spender = 0                  # Max payment store
print('Name    Amount  Discount  Payable  No of orders')  # Table heading

for i in orders:                     # Har order loop
    amount = 0                       # Total bill reset
    n = 0                            # Items count reset

    for key, value in i['items'].items():   # Items loop
        amount += menu[key] * value         # Bill add
        n += value                          # Quantity add
        # print(i['name'], key, value)      # Debug line

    if amount >= 300:                # 300 se zyada
        discount = amount * 0.2      # 20% discount
        pay = amount - discount      # Final payment

    elif amount >= 200 and amount <= 299:   # 200–299 range
        discount = amount * 0.1      # 10% discount
        pay = amount - discount      # Final payment

    else:                            # 200 se kam
        discount = amount * 0        # No discount
        pay = amount - discount      # Same amount

    if pay >= highest_spender:       # Max check
        highest_spender = pay        # New highest
        highest_spender_name = i['name']  # Name store

    print(i['name'], "  ", int(amount), "     ", int(discount),
          "      ", int(pay), "           ", n)   # Row print

print("name of highest_spender_name", highest_spender_name, highest_spender)
# Highest spender print
```
